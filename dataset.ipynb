```python
import kagglehub
import pandas as pd
import os

datasets = {
    "laptops": "mayankkumarpoddar/smartprix-laptops-dataset",
    "mice": "ellimaaac/gaming-mouse-specs",
    "headphones": "maulikgajera/global-headphones-specifications-and-market-dataset",
    "monitors": "durjoychandrapaul/amazon-products-sales-monitor-dataset",
    "hardware": "dilshaansandhu/general-computer-hardware-dataset"
}

documents = []

for category, dataset_id in datasets.items():

    print(f"\nDownloading {category}...")
    dataset_path = kagglehub.dataset_download(dataset_id)

    for root, dirs, files in os.walk(dataset_path):

        for file in files:

            if file.endswith(".csv"):

                csv_path = os.path.join(root, file)
                print("Loading:", csv_path)

                try:
                    df = pd.read_csv(csv_path)

                    for _, row in df.iterrows():

                        text = f"CATEGORY: {category}\n"

                        for col in df.columns:

                            value = row[col]

                            if pd.notna(value):
                                text += f"{col}: {value}\n"

                        documents.append(text)

                except Exception as e:
                    print(f"Error loading {csv_path}: {e}")

print(f"\nTotal documents: {len(documents)}")
print("\nSample document:\n")
print(documents[0][:1000])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_texts(
    documents,
    embedding=embeddings
)

vectorstore.save_local("tech_products_db")

print("FAISS database saved.")
```
